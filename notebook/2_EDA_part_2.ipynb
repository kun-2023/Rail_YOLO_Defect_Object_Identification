```python
from pathlib import Path
```

```python
current_dir=Path.cwd()
print(current_dir)
```

```python
if current_dir.name=="notebook":
    project_root=current_dir.parent
else:
    project_root=current_dir
data_dir=project_root/"data"
print(project_root)
print(data_dir)
```

```python
for folder in ["train", "test", "valid"]:
    image_folder=data_dir/folder/"images"
    extensions=sorted({
        file.suffix.lower() for file in image_folder.iterdir() if file.is_file()
    })

    print(f"The image type in {folder} is {extensions}.")
```
